```python
df.groupby('category').agg({'amt':'mean','is_fraud':'mean'}).sort_values(by='is_fraud', ascending=False)
```

```python
#Column Chart
plt.figure(figsize=(6,4))
sns.countplot(x='is_fraud', data=df,color='orange')
plt.xlabel("Fraud Status (0 = Non-Fraud, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.title("Distribution of Fraud vs Non-Fraud Transactions")
plt.show()
```

Fraudulent transactions represent a very small fraction of total transactions, confirming that the dataset is highly imbalanced.

```python
plt.figure(figsize=(10,6))
sns.barplot(x='amt', y='category', data=df, estimator='mean' , color='orange')
plt.xlabel("Average Transaction Amount")
plt.ylabel("Transaction Category")
plt.title("Average Transaction Amount by Category")
plt.show()
```

Certain categories (especially online shopping categories) tend to have higher average transaction amounts.

```python
#Bar Chart
fraud_rate = df.groupby('category')['is_fraud'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(x='is_fraud', y='category', data=fraud_rate,color='orange')
plt.xlabel("Fraud Rate")
plt.ylabel("Transaction Category")
plt.title("Fraud Rate by Category")
plt.show()
```

Online transaction categories show higher fraud rates compared to physical store transactions.

```python
#Pie Chart
fraud_by_gender = df.groupby('gender')['is_fraud'].mean().reset_index()
plt.figure(figsize=(6, 6))
plt.pie(fraud_by_gender['is_fraud'],labels=fraud_by_gender['gender'],autopct='%1.1f%%',startangle=90,shadow=True,explode=[0.05] * len(fraud_by_gender))
plt.title('Fraud Rate by Gender')
```

This pie chart visualizes the distribution of fraud rates across different genders,
helping to identify if fraud is more prevalent in any particular gender group.

```python
#Line Chart
df['month'] = df['trans_date_trans_time'].dt.month
plt.figure(figsize=(8,5))
sns.lineplot(x='month', y = 'amt' ,ci=None , data=df, color='orange')
plt.xlabel("Month")
plt.ylabel("Transactions")
plt.title("Monthly Transaction Volume")
plt.show()
```

Reveals seasonal spending patterns throughout the year
& Identifies months with highest and lowest transaction volumes

```python
#Scatter Plot
monthly_counts = df.groupby('month').size().reset_index(name='transaction_count')

plt.figure(figsize=(8,5))
sns.scatterplot(x='month', y='transaction_count', data=monthly_counts, s=100 ,color='orange')
plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.title("Monthly Transaction Volume (Scatter Plot)")
plt.show()
```

This scatter plot highlights how transaction volume varies across months. It helps identify months with unusually high or low activity.

```python
#Histogram
df['age'] = 2026 - df['dob'].dt.year
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='age', color='orange', bins=30, kde=True)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution by Fraud Status")
plt.show()
```

This shows whether fraud is concentrated in specific age groups.

```python
#Heatmap
numeric_cols = df.select_dtypes(include=['int64','float64'])
correlation_matrix = numeric_cols.corr()
plt.figure(figsize=(12,8))
sns.heatmap(correlation_matrix,annot=True,cmap='coolwarm')
plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```
